```python
import pandas as pd
import glob
import os

input_path = 'hospital_data/raw/*.csv' # Path to your folder
output_folder = 'hospital_data/cleaned_data/'

os.makedirs(output_folder, exist_ok=True)

files = glob.glob(input_path)

for file in files:
    file_name = os.path.basename(file)
    df = pd.read_csv(file)
    df = df.drop_duplicates()
    df.columns = df.columns.str.strip().str.lower() # Standardize headers
    df = df.fillna("Unknown") # Handle missing values
    df.to_csv(f"{output_folder}cleaned_{file_name}", index=False)
    print(f"Processed: {file_name}")
```
